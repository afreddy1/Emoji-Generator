```python
import pandas as pd
import emoji
import re
from tqdm.notebook import tqdm as tqdm
```

```python

df = pd.read_csv("tweets1.csv") 

# extracts emojis
def extract_emojis(text):
    return [char for char in text if char in emoji.EMOJI_DATA]

# new 'emojis' column
df['emojis_present'] = df['Text'].apply(extract_emojis)
```

```python
df
```

```python
from itertools import chain

all_emojis = list(set(chain.from_iterable(df['emojis_present'])))
all_emojis.sort()  #for consistency
emoji_dict = {e: i for i, e in enumerate(all_emojis)}
```

```python
emoji_dict
```

```python
import numpy as np

def multi_hot_encode(emoji_list):
    vec = np.zeros(len(emoji_dict), dtype=int)
    for e in emoji_list:
        if e in emoji_dict:
            vec[emoji_dict[e]] = 1
    return vec

df['emojis'] = df['emojis_present'].apply(multi_hot_encode)
```

```python
df
```

```python
# remove emojis from text (tweet)
def remove_emojis(text):
    return ''.join(char for char in text if not emoji.is_emoji(char))


df['Text'] = df['Text'].apply(remove_emojis)
```

```python
df
```

```python
from transformers import pipeline

classifier = pipeline("text-classification", model="j-hartmann/emotion-english-distilroberta-base", top_k=None)
```

```python

df2 = pd.read_csv("tweets1.csv")  # Replace with your actual dataframe

# Function to get emotion probabilities for each tweet
def get_emotion_probs(text):
    emotion_scores = classifier(text)
    # Extract only the probabilities for each emotion
    return {score['label']: score['score'] for score in emotion_scores[0]}

tqdm.pandas()
# Apply function to the 'Text' column and create new columns for each emotion
emotion_columns = ['anger', 'disgust', 'fear', 'joy', 'neutral', 'sadness', 'surprise']
for emotion in tqdm(emotion_columns):
    df2[emotion] = df2['Text'].progress_apply(lambda x: get_emotion_probs(x).get(emotion, 0))

# Preview the dataframe
print(df2.head())
```

```python
# Get the first tweet
first_tweet = df['Text'].iloc[5049]

# Run emotion classifier on that tweet
emotion_scores = classifier(first_tweet)

# Format the output
emotion_dict = {score['label']: score['score'] for score in emotion_scores[0]}

# Print results
print(f"Tweet: {first_tweet}\n")
print("Emotion scores:")
for emotion, score in emotion_dict.items():
    print(f"{emotion}: {score:.4f}")
```

```python
from transformers import AutoModelForSequenceClassification, AutoTokenizer, AutoConfig
from scipy.special import softmax
import torch

# pipeline setup
MODEL = "cardiffnlp/twitter-roberta-base-sentiment-latest"
tokenizer = AutoTokenizer.from_pretrained(MODEL)
config = AutoConfig.from_pretrained(MODEL)
model = AutoModelForSequenceClassification.from_pretrained(MODEL)


def preprocess(text):
    new_text = []
    for t in text.split(" "):
        t = '@user' if t.startswith('@') and len(t) > 1 else t
        t = 'http' if t.startswith('http') else t
        new_text.append(t)
    return " ".join(new_text)

# Sentiment probability extractor
def get_sentiment_probs(text):
    text = preprocess(text)
    encoded_input = tokenizer(text, return_tensors='pt', truncation=True, max_length=512)
    with torch.no_grad():
        output = model(**encoded_input)
    scores = output.logits[0].numpy()
    scores = softmax(scores)
    return {
        'negative': float(scores[0]),
        'neutral': float(scores[1]),
        'positive': float(scores[2])
    }

# applyyy
from tqdm import tqdm
tqdm.pandas()

df_with_emoji_emotion['negative'] = df_with_emoji_emotion['Text'].progress_apply(lambda x: get_sentiment_probs(x)['negative'])
df_with_emoji_emotion['neutral_sent'] = df_with_emoji_emotion['Text'].progress_apply(lambda x: get_sentiment_probs(x)['neutral'])
df_with_emoji_emotion['positive'] = df_with_emoji_emotion['Text'].progress_apply(lambda x: get_sentiment_probs(x)['positive'])


print(df_with_emoji_emotion.head())
```
