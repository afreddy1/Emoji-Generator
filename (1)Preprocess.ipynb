```python
import pandas as pd
import os

df_list = []
error_files = []

for filename in os.listdir("archive"):
    if filename.endswith(".csv"):
        file_path = os.path.join("archive", filename)
        try:
            # Try reading with utf-8 encoding
            df = pd.read_csv(file_path, encoding='utf-8')
        except UnicodeDecodeError:
            # Try ISO encoding if utf-8 fails
            try:
                df = pd.read_csv(file_path, encoding='ISO-8859-1')
            except Exception as e:
                print(f" Could not read {filename}: {e}")
                error_files.append(filename)
                continue
        except Exception as e:
            print(f" Could not read {filename}: {e}")
            error_files.append(filename)
            continue

        # Only keep "Text" column if it exists
        if "Text" in df.columns:
            df_list.append(df[["Text"]])
        else:
            print(f"⚠️ 'Text' column not found in {filename}")
            error_files.append(filename)

# combines all valid DataFrames
combined_df = pd.concat(df_list, ignore_index=True)

# drops NaNs and empty strings
combined_df.dropna(subset=["Text"], inplace=True)
combined_df = combined_df[combined_df["Text"].str.strip() != ""]

# save to new CSV
combined_df.to_csv("tweets0.csv", index=False)

if error_files:
    print("\n skipped files ")
    for ef in error_files:
        print(f" - {ef}")

```

```python
full_df = full_df.dropna(subset=['Text'])
```

```python
# true if tweet consists only of special characters or whitespace
def is_special_chars_only(tweet):
    # Return True if tweet consists only of special characters or whitespace
    return not bool(re.search(r'[a-zA-Z0-9]', tweet))

full_df = full_df[~full_df['Text'].apply(is_special_chars_only)]
```

```python
# checks if tweet contains only URLs or mentions
def is_url_or_mention(tweet):
    return re.match(r"(http\S+|@\w+)+", tweet) is not None

full_df = full_df[~full_df['Text'].apply(is_url_or_mention)]
```

```python
# true if tweet consists only of special characters or whitespace
def is_special_chars_only(tweet):
    return not bool(re.search(r'[a-zA-Z0-9]', tweet))

full_df = full_df[~full_df['Text'].apply(is_special_chars_only)]
```

```python
max_length = 150  # max length of tweet
full_df = full_df[full_df['Text'].apply(lambda x: len(x) <= max_length)]
```

```python
full_df = full_df[~full_df["Text"].str.contains(r"\n|\t")]
```

```python
full_df = full_df[~full_df["Text"].str.contains("http", case=False, na=False)]
```

```python
def has_text_chars(tweet):
    return bool(re.search(r'[a-zA-Z0-9]', tweet))

full_df = full_df[full_df["Text"].apply(has_text_chars)]
```

```python
def extract_emojis(text):
    return [char for char in str(text) if emoji.is_emoji(char)]

all_emojis = full_df['Text'].dropna().apply(extract_emojis)

flat_emojis = set([emoji for sublist in all_emojis for emoji in sublist])

print(f"Number of unique emojis: {len(flat_emojis)}")
print(flat_emojis)
```
