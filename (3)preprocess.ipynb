```python
import ast
import pandas as pd


df['emojis_present'] = df['emojis_present'].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else x)

unique_emoji_lists = df['emojis_present'].apply(lambda emojis: set(emojis))
emoji_counts = unique_emoji_lists.explode().value_counts()

most_common_emoji = emoji_counts.idxmax()
most_common_count = emoji_counts.max()
most_common_probability = most_common_count / emoji_counts.sum()

print(f"Most Common Emoji: {most_common_emoji}, Count: {most_common_count}")
print(f"Most Common Emoji Probability: {most_common_probability}")
```
